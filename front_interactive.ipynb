#### Run the first cell, select continent and date in the interactive dropdown menus, then run the second cell

```python
#setup
from pathlib import Path
import sys
scriptspath = Path.cwd() / 'scripts'
scriptspath = scriptspath.as_posix()
if scriptspath not in sys.path:
    sys.path.append(scriptspath)
from wf_functions import get_continents, get_dayrangeoptions, get_map

import ipywidgets as widgets
from IPython.display import display
import pandas as pd


continents = get_continents()
dayrangeoptions = get_dayrangeoptions()

# Create dropdown with default selection
continents_dropdown = widgets.Dropdown(
    options=continents,
    value="Choose continent",  # Set default value
    description='Continent:',
    disabled=False,
)
display(continents_dropdown)

#create dayrange selector
dayrange_dropdown = widgets.Dropdown(
    options=dayrangeoptions,
    value=1,  # Set default value
    description='Day Range:',
    disabled=False,
)
display(dayrange_dropdown)

#create a date picker 
date_picker = widgets.DatePicker(
    description = 'Pick a Date', 
    disabled = False
)
display(date_picker)
```

#### Run the next cell only after you selected continent and date.

```python
#do not change anything here
get_map(continents_dropdown.value, date_picker.value, dayrange_dropdown.value)
```
