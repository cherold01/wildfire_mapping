```python
import pandas as pd
import requests
import geopandas as gpd
import folium
import numpy as np
```

```python
countries = gpd.read_file('data/geoBoundariesCGAZ_ADM0.gpkg').to_crs(4326)
countries.info()
countries.crs
```

```python
countries['shapeName']
```

```python
countries_simple = countries.copy()
countries_simple = countries_simple.drop(columns = ['id', 'shapeGroup', 'shapeType'])
countries_simple.rename(columns={'shapeName': "CountryName"}, inplace=True)
countries_simple.geometry = countries.geometry.simplify(tolerance = 0.01) #simplify to 0.01 degree, roughly 1km
countries_simple.info()
```

```python
countries_simple.to_file('data/countries_simple.gpkg')
```
